# Búsqueda de números pares en una lista
Verificar que este tipo de búsqueda tiene una complejidad $O(n)$.

```python
import random                   #---Para genera números enteros y listas de enteros aleatorias
import time                     #---Para medir el tiempo de ejecución
import matplotlib.pyplot as plt #--- Permite realizar gráficos
```

```python
#--------------------------Parámetros
n0,n = 10, 10000                    # Tamaño inicial de la lista--- Tamaño máximo de la lista
k,p = 0,0                     # Contador de procesos
#--------------------------Listas para almacenar resultados
Tiempo = []               # Almacena los tiempos de ejecución
Dimensione = []               # Almacena los tamaños de las listas
```

```python
#--------------------------------------------Ciclo principal-----------------------
while n0 < n:
    N = random.sample(range(1, 10 * n), n0) #--Generar una lista aleatoria de n0 enteros diferentes
    inicio = time.perf_counter()            #--Iniciar el cronómetro para medir el tiempo de ejecusión
    #-----------------------------------------Buscar y contar los números pares
    pares,i = 0,0
    while i < n0:
        if N[i] % 2 == 0:
            pares += 1
        i += 1
    #-----------------------------------------Finalizar el cronómetro
    fin = time.perf_counter()
    tiempo = fin - inicio                  #---Calcula el tiempo de ejecución
    #--------------------------------------Guardar resultados
    Tiempo.append(tiempo)
    Dimensione.append(n0)
    p += pares                            #---Acumular el número de pares
    k += 1                                #---Incrementar el contador
    n0 += 10                              #---Incrementar el tamaño de la lista
```

# **Gráfica**

```python
tiempos_ms = [t * 1000 for t in Tiempo] # Convertir los tiempos de segundos a milisegundos
#-----------Gráfica
plt.plot(Dimensione, tiempos_ms, 'o-', linewidth=1.5, markersize=4)
plt.xlabel(r"$n$")
plt.ylabel(r"$t\,(ms)$")
plt.title("Complejidad temporal: O(n)")
plt.xscale('log')
plt.yscale('log')
plt.grid(True)
plt.show()
```
